## Issue 1: Gower tie-breaker treats categorical targets as numerical

```python
import pickle
import pandas as pd
import networkx as nx
import pickle as pkl
import matplotlib.pyplot as plt


# Open the pickle file in binary read mode ('rb')
with open('graph_349519_OGBG.pkl', 'rb') as file:
    data = pickle.load(file)
```

```python
import sys
from pathlib import Path

project_root = Path.cwd().parent 
sys.path.insert(0, str(project_root))

from algorithm.methods import ranking, significance_test, find_groups
```

```python
edge_index = data['edge_index']
num_nodes = data['num_nodes']

G = nx.Graph()

G.add_nodes_from(range(num_nodes))

edges = list(zip(edge_index[0], edge_index[1]))
G.add_edges_from(edges)
```

```python
attributes = data['node_feat']
lu = pd.DataFrame(attributes)
lu['target'] = lu[0] >= 6

categorical_cols = ["1", "7", "8", "9"]
columns_to_ignore = [0, 'target']

for col in lu.columns:
    if col not in categorical_cols + columns_to_ignore:
        lu[col] = lu[col].astype(float)
```

```python
result = find_groups(G, 20, lu, 0, columns_to_ignore, categorical_cols, ablation_mode=False, use_multiprocessing=False)
```
