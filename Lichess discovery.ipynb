```python
import pandas as pd
import networkx as nx

edges = pd.read_csv("data/processed/edges_snowball_core.csv")
```

```python
edges
```

```python
players = pd.read_csv("data/processed/players_snowball.csv").set_index("username")
players
```

```python
graph = nx.Graph()
graph.add_nodes_from(players.index)
for row in edges.itertuples():
    graph.add_edge(row.player_a, row.player_b, weight=row.distance)
```

```python
lu = players.copy()

categorical_cols = ['title', 'main_tc']
columns_to_ignore = [
    'target_growth_games', 'target_growth_1m', 'target_growth_1m_top20', 'target_growth_3m', 'target_growth_3m_top20', 
    'target_growth_6m', 'target_growth_6m_top20', 'rating', 'username']


for col in lu.columns:
    if col not in categorical_cols + columns_to_ignore:
        lu[col] = lu[col].astype(float)
```

```python
lu
```

```python
import sys
from pathlib import Path

project_root = Path.cwd().parent 
sys.path.insert(0, str(project_root))

from algorithm.methods import ranking, significance_test, find_groups

result = find_groups(
    graph, 20, lu, 
    'target_growth_games', 
    columns_to_ignore, 
    categorical_cols, ablation_mode=False, use_multiprocessing=False)
```

```python
result
```

```python
reference = result.iloc[0]['reference']
subgroup = result.iloc[0]['subgroup']

print("Avg target of the reference: ", lu.loc[lu.index.isin(reference), 'target_growth_games'].mean())
print("Avg target of the subgroup: ", lu.loc[lu.index.isin(subgroup), 'target_growth_games'].mean())
```
